```python
import numpy as np

a = np.array([1,2,3])

b = np.array([4,5,6])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**3) #exponent
print(np.sqrt(a)) # to find the square root
print(np.power(a,5)) # for n exponent
arr = np.array([10,20,20,30,40,50]) # defining 2d array
print(np.mean(arr)) # to find the average
print(np.median(arr)) # to find the median
print(np.max(arr)) # to find the max value
print(np.min(arr)) # to find the min value
print(np.sum(arr)) # total value
print(np.std(arr)) # standard deviation
print(np.var(arr)) # variance
arr2 = np.array([1,2,3,4,5,6])
print(arr2.reshape(2,3))



```
